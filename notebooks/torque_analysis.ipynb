```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/home/robot/workspaces/robot_rodeo_gym_ws/logs/train_ftr_11001809/attempt_0/torque.csv')

flippers = ['FL', 'FR', 'RL', 'RR']

mean_cols = [f'torque/flipper_torque_{f}' for f in flippers]
max_cols  = [f'torque/flipper_torque_{f}_max' for f in flippers]
min_cols  = [f'torque/flipper_torque_{f}_min' for f in flippers]

mean_mean = df[mean_cols].mean(axis=1)
mean_max  = df[max_cols].mean(axis=1)
mean_min  = df[min_cols].mean(axis=1)

x = df['collected_frames']

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, mean_mean, label='mean torque')
ax.plot(x, mean_max,  label='max torque')
ax.plot(x, mean_min,  label='min torque')
ax.set_xlabel('Collected frames')
ax.set_ylabel('Torque (Nm)')
ax.grid(True,'major')
ax.set_title('Flipper torque across training (mean over FL/FR/RL/RR)')
ax.legend()
plt.tight_layout()
plt.show()
```
